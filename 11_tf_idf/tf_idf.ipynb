```python
from sklearn.feature_extraction.text import TfidfVectorizer

corpus = [
    "Thor eating pizza, Loki is eating pizza, Ironman ate pizza already",
    "Apple is announcing new iphone tomorrow",
    "Tesla is announcing new model-3 tomorrow",
    "Google is announcing new pixel-6 tomorrow",
    "Microsoft is announcing new surface tomorrow",
    "Amazon is announcing new eco-dot tomorrow",
    "I am eating biryani and you are eating grapes"
]
```

```python
v = TfidfVectorizer()
v.fit(corpus)
transform_output = v.transform(corpus)
```

```python
v.vocabulary_
```

```python
all_features_names = v.get_feature_names_out()

for word in all_features_names:
    index = v.vocabulary_.get(word)
    idf_score = v.idf_[index]
    print(f"{word}: {idf_score}")
```

```python
import pandas as pd

df = pd.read_csv('Ecommerce_data.csv')
print(df.shape)
```

```python
df.head()
```

```python
df['label'].value_counts()
```

```python
df['label_num'] = df['label'].map({
    'Household' : 0, 
    'Books': 1, 
    'Electronics': 2, 
    'Clothing & Accessories': 3
})

df.head()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    df.Text,
    df.label_num,
    test_size=0.2,
    random_state=2022,
    stratify=df.label_num
)
```

```python
print(f"Shape of XTrain {X_train.shape}")
print(f"Shape of Xtest {X_test.shape}")
```

```python
X_train.head()
```

```python
y_train.value_counts()
```

```python
y_test.value_counts()
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report
from sklearn.pipeline import Pipeline

clf = Pipeline([
    ('vec_tfidf',TfidfVectorizer()),
    ('knn',KNeighborsClassifier())
])

clf.fit(X_train,y_train)

y_pred= clf.predict(X_test)

print(classification_report(y_test,y_pred))
```

```python
from sklearn.naive_bayes import MultinomialNB

clf = Pipeline([
    ('vec_tfidf',TfidfVectorizer()),
    ('multi_nb',MultinomialNB())
])

clf.fit(X_train,y_train)

y_pred= clf.predict(X_test)

print(classification_report(y_test,y_pred))
```

```python
from sklearn.ensemble import RandomForestClassifier

clf = Pipeline([
    ('tfidf',TfidfVectorizer()),
    ('randforestclass',RandomForestClassifier()),
])

clf.fit(X_train,y_train)

y_pred= clf.predict(X_test)

print(classification_report(y_test,y_pred))
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")

def preProcess(text):
    doc = nlp(text)
    filtered_tokens = []
    for token in doc:
        if token.is_stop or token.is_punct:
            continue
        filtered_tokens.append(token.lemma_)
    
    return " ".join(filtered_tokens)
```

```python
df['prepreocess_text']= df['Text'].apply(preProcess)
```

```python
df.head()
```

```python
df.Text[0]
```

```python
df.prepreocess_text[0]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    df.prepreocess_text,
    df.label_num,
    test_size=0.2,
    random_state=2022,
    stratify=df.label_num
)
```

```python
clf = Pipeline([
    ('tfidf',TfidfVectorizer()),
    ('rand_forest',RandomForestClassifier())
])

clf.fit(X_train,y_train)

y_pred = clf.predict(X_test)

print(classification_report(y_test,y_pred))
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test,y_pred)
cm
```

```python
from matplotlib import pyplot as plt
import seaborn as sn

plt.figure(figsize=(10,7))
sn.heatmap(cm,annot=True,fmt='d')
plt.xlabel('Prediction')
plt.ylabel('Truth')
```
